```python
def power_family(z, lam):
    # Monotone-increasing for z>0
    if np.isclose(lam, 0.0):
        return np.log(z)
    else:
        return (np.power(z, lam) - 1.0) / lam

lam_grid = np.linspace(-2, 2, 161)  # coarse grid; refine around the peak later
best = (-np.inf, None, None)        # (r, lam_x, lam_y)

x=np.abs( var1x ).flatten()
y=np.abs( var2x ).flatten()

for lx in lam_grid:
    zx = power_family(x, lx)
    zx = (zx - zx.mean())/zx.std(ddof=1)
    for ly in lam_grid:
        zy = power_family(y, ly)
        zy = (zy - zy.mean())/zy.std(ddof=1)
        r = np.corrcoef(zx, zy)[0,1]
        if r > best[0]:
            best = (r, lx, ly)

r_max, lam_x_star, lam_y_star = best
print(f"Maximized correlation r = {r_max:.3f} at λ_x={lam_x_star:.3f}, λ_y={lam_y_star:.3f}")
```

```python
from sklearn.preprocessing import PowerTransformer

def yeo_johnson_1d(z, lam):
    # Use sklearn's implementation by fitting a dummy transformer with forced lambda
    pt = PowerTransformer(method='yeo-johnson', standardize=False)
    pt.lambdas_ = np.array([lam], dtype=float)  # inject λ
    # sklearn expects to call fit before transform; emulate fitted state:
    # We'll just use its internal formula via transform after setting lambdas_
    zt = PowerTransformer(method='yeo-johnson', standardize=False)
    zt.lambdas_ = np.array([lam], dtype=float)
    # HACK: zt.transform calls yeo-johnson using zt.lambdas_
    return zt.transform(z.reshape(-1,1)).ravel()

lam_grid = np.linspace(-4, 4, 161)
best = (-np.inf, None, None)
var1x=copy.deepcopy(var1[:,0:101])
var2x=copy.deepcopy(var2[:,0:101])

x=np.abs( var1x ).flatten()
y=np.abs( var2x ).flatten()

for lx in lam_grid:
    zx = yeo_johnson_1d(x, lx)
    zx = (zx - zx.mean())/zx.std(ddof=1)
    for ly in lam_grid:
        zy = yeo_johnson_1d(y, ly)
        zy = (zy - zy.mean())/zy.std(ddof=1)
        r = np.corrcoef(zx, zy)[0,1]
        if r > best[0]:
            best = (r, lx, ly)

r_max, lam_x_star, lam_y_star = best
print(f"Maximized correlation r = {r_max:.3f} at λ_x={lam_x_star:.3f}, λ_y={lam_y_star:.3f}")
```
