```python
year=2004
months=[6,7,8] #None

do_objs=[
        'MCS_Tb_Objects', 
        'MRG_Objects' ,
            'IGW_Objects', 
            'Kelvin_Objects', 
            'EIG0_Objects', 
            'ER_Objects' ]

do_objs=[
        'MCS_Tb_Objects', 
        'MRG_Objects', 
        'IGW_Objects', 
        'Kelvin_Objects', 
        'EIG0_Objects', 
        'ER_Objects',
        'FR_Objects', 
        'CY_Objects', 
        'ACY_Objects', 
        'MS_Objects', 
        'IVT_Objects', 
        'AR_Objects', 
        'CY_z500_Objects', 
        'ACY_z500_Objects', 
        'COL_Objects', 
        'JET_Objects', 
        'BT_Objects', 
        'TC_Objects', 
        ]
do_objs=[
        #'MCS_Tb_Objects', 
        'MRG_Objects', 
        'ER_Objects',
        'Kelvin_Objects', 
        'EIG0_Objects', 
        'IGW_Objects', 
       ]
do_objs=[
        'MCS_Tb_Objects', 
       ]
"""
do_objs=[
        'Kelvin_Objects', 'ER_Objects', 'MRG_Objects',
       ]
"""
Al =[]
for objn in do_objs:
    
    A = { 
        'objName': objn ,
        'year':year , 'months':months ,
            }

    Xo = nc.rd_objprec(year=2004,exp='ERA5_Obs',objName=objn,months=months,roll=True)
    Xh = nc.rd_objprec(year=2004,exp='CAM7_HR', objName=objn,months=months)
    XL = nc.rd_objprec(year=2004,exp='CAM7_LR', objName=objn,months=months)
    
    ndo,nyo,nxo=Xo.sizes['time'],Xo.sizes['yc'],Xo.sizes['xc']
    ndc,nyc,nxc=Xh.sizes['time'],Xh.sizes['yc'],Xh.sizes['xc']
    ndL,nyL,nxL=XL.sizes['time'],XL.sizes['yc'],XL.sizes['xc']
    
    latL=XL.lat.values
    lonL=XL.lon.values
    lato=Xo.lat.values
    lono=Xo.lon.values
    lath=Xh.lat.values
    lonh=Xh.lon.values
    
    objprec_h = Xh[f'{objn}-PR'].values
    objprec_o = Xo[f'{objn}-PR'].values
    objprec_L = XL[f'{objn}-PR'].values

    objfreq_h = Xh[f'{objn}-freq'].values
    objfreq_o = Xo[f'{objn}-freq'].values
    objfreq_L = XL[f'{objn}-freq'].values

    prec_h    = Xh[f'PR'].values
    prec_o    = Xo[f'PR'].values
    prec_L    = XL[f'PR'].values
    
    objprec_h_tv=np.average( objprec_h, axis=0 )
    objprec_o_tv=np.average( objprec_o, axis=0 )
    objprec_L_tv=np.average( objprec_L, axis=0 )

    objfreq_h_tv=np.average( objfreq_h, axis=0 )
    objfreq_o_tv=np.average( objfreq_o, axis=0 )
    objfreq_L_tv=np.average( objfreq_L, axis=0 )

    prec_h_tv=np.average( prec_h, axis=0 )
    prec_o_tv=np.average( prec_o, axis=0 )
    prec_L_tv=np.average( prec_L, axis=0 )

    A['lono']=lono
    A['lato']=lato
    A['lonh']=lonh
    A['lath']=lath
    A['lonL']=lonL
    A['latL']=latL

    #A['objprec_o'] = objprec_o
    #A['objprec_h'] = objprec_h

    A['objfreq_L_tv'] = objfreq_L_tv
    A['objfreq_o_tv'] = objfreq_o_tv
    A['objfreq_h_tv'] = objfreq_h_tv

    A['objprec_L_tv'] = objprec_L_tv
    A['objprec_o_tv'] = objprec_o_tv
    A['objprec_h_tv'] = objprec_h_tv

    A['objIprec_L_tv'] = objprec_L_tv / (objfreq_L_tv+1.e-6)
    A['objIprec_o_tv'] = objprec_o_tv / (objfreq_o_tv+1.e-6)
    A['objIprec_h_tv'] = objprec_h_tv / (objfreq_h_tv+1.e-6)

    A['objFprec_L_tv'] = objprec_L_tv / (prec_L_tv+1.e-6)
    A['objFprec_o_tv'] = objprec_o_tv / (prec_o_tv+1.e-6)
    A['objFprec_h_tv'] = objprec_h_tv / (prec_h_tv+1.e-6)

    A['prec_L_tv'] = prec_L_tv 
    A['prec_o_tv'] = prec_o_tv 
    A['prec_h_tv'] = prec_h_tv 

    Al.append(A)
```
