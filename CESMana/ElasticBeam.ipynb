```python
import numpy as np
from scipy.optimize import fsolve

import matplotlib.pyplot as plt

def find_catenary_length(b, h):
    # Equation to solve: h = a * (cosh(b/(2a)) - 1)
    def equation(a):
        return a * (np.cosh(b / (2*a)) - 1) - h

    # Initial guess for a
    a0 = h

    # Solve for a
    a_solution = fsolve(equation, a0)[0]

    # Compute length
    L = 2 * a_solution * np.sinh(b / (2 * a_solution))
    return L

# Example usage:
b = 3.0  # meters between stakes
h = 1.0  # height of arc

length = find_catenary_length(b, h)
print(f"Catenary length: {length:.4f} meters")


a=.2
x=np.linspace( -1, 1, 100)
caten= -a* np.cosh( x/a )
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import ellipj

# Parameters
a = 1.0          # scaling factor
k = 0.9          # elliptic modulus
s = np.linspace(-2*a, 2*a, 1000)  # arc-length parameter

# Compute sn(u, k)
u = s / a
sn, _, _, _ = ellipj(u, k**2)  # note: scipy uses m = k^2

# Compute theta(s)
theta = 2 * np.arcsin(k * sn)

# Compute dx/ds and dy/ds
dxds = np.cos(theta)
dyds = np.sin(theta)

# Custom trapezoidal integration (works like cumtrapz)
def trapz_integrate(y, x):
    return np.concatenate([[0], np.cumsum((y[1:] + y[:-1]) * np.diff(x) / 2)])

# Integrate to get x(s), y(s)
x = trapz_integrate(dxds, s)
y = trapz_integrate(dyds, s)

# Plot the elastica curve
plt.figure(figsize=(8, 6))
plt.plot(x, y, label="Elastica curve")
plt.axis('equal')
plt.xlabel("x")
plt.ylabel("y")
plt.title("Euler Elastica Curve from θ(s)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np

def compute_curve_length(x, y):
    """
    Compute the arc length of a 2D curve given x and y coordinates.
    
    Parameters:
    - x, y: arrays of same length containing curve coordinates
    
    Returns:
    - Total arc length (float)
    """
    x = np.asarray(x)
    y = np.asarray(y)
    dx = np.diff(x)
    dy = np.diff(y)
    ds = np.sqrt(dx**2 + dy**2)
    return np.sum(ds)
```
